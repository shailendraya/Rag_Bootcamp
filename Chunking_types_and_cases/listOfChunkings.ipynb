# Chunking type and thier use cases

|  # | Chunking Strategy            | Best For              | Example Input           | Real-world Use Case                 |
| -: | ---------------------------- | --------------------- | ----------------------- | ----------------------------------- |
|  1 | Character Chunking           | Plain text            | `.txt`                  | Small text files, demos             |
|  2 | Recursive Character Chunking | General documents     | `.txt`, `.pdf`, `.docx` | General-purpose RAG                 |
|  3 | Token Chunking               | LLM optimization      | Any text                | OpenAI, Claude, Gemini token limits |
|  4 | Sentence Chunking            | FAQs                  | FAQ text                | Question answering                  |
|  5 | Paragraph Chunking           | Articles              | Blog, news              | Knowledge base                      |
|  6 | Semantic Chunking            | Conceptual text       | Tutorials, books        | High-quality RAG                    |
|  7 | Threshold Semantic Chunking  | Long technical docs   | AI/ML docs              | Production semantic search          |
|  8 | Markdown Chunking            | Documentation         | `.md`                   | GitHub, LangChain docs              |
|  9 | HTML Chunking                | Websites              | `.html`                 | Website chatbot                     |
| 10 | JSON Chunking                | APIs                  | `.json`                 | API documentation                   |
| 11 | XML Chunking                 | Config files          | `.xml`                  | Enterprise systems                  |
| 12 | CSV Chunking                 | Structured data       | `.csv`                  | Analytics assistant                 |
| 13 | Table Chunking               | Tables                | PDF tables              | Financial reports                   |
| 14 | Code Chunking                | Source code           | `.py`, `.js`, `.java`   | GitHub Copilot-like apps            |
| 15 | Notebook Chunking            | Jupyter notebooks     | `.ipynb`                | AI tutorial search                  |
| 16 | Parent-Child Chunking        | Large documents       | Books                   | ParentDocumentRetriever             |
| 17 | Contextual Chunking          | Long reports          | Annual reports          | Better context retention            |
| 18 | Sliding Window Chunking      | Sequential text       | Books                   | Preserve continuity                 |
| 19 | Hierarchical Chunking        | Books/manuals         | Book PDF                | Multi-level retrieval               |
| 20 | Hybrid Chunking              | Mixed documents       | PDF + Markdown          | Enterprise RAG                      |
| 21 | Graph Chunking               | Entity-rich documents | Contracts, Wikipedia    | Graph RAG                           |
| 22 | LLM-based Chunking           | Complex documents     | Research papers         | Intelligent chunk boundaries        |
| 23 | Agentic Chunking             | Dynamic corpora       | Mixed enterprise docs   | Agentic RAG                         |
| 24 | Multi-modal Chunking         | Text + Images         | PDF with figures        | Vision RAG                          |
| 25 | Layout-aware Chunking        | Complex PDFs          | Invoices, resumes       | Document AI                         |


## Production Recommendation

| Document Type              | Recommended Strategy                 |
| -------------------------- | ------------------------------------ |
| Plain text                 | Recursive + Semantic                 |
| Technical documentation    | Markdown + Semantic + Token          |
| PDFs                       | Layout-aware + Semantic              |
| Source code                | Code Chunking                        |
| Books                      | Parent-Child + Sliding Window        |
| Research papers            | Layout-aware + Semantic + Contextual |
| Enterprise knowledge base  | Hybrid Chunking                      |
| Contracts                  | Layout-aware + Graph + Contextual    |
| API documentation          | JSON + Markdown                      |
| GitHub repositories        | Markdown + Code Chunking             |
| Mixed enterprise documents | Hybrid + Agentic                     |
